<h3> 1.Write a Python function safe_divide(a, b) that returns the result of a divided by b, and handles ZeroDivisionError by returning the string 'Cannot divide by zero'.


```python
def safe_divide(a, b):
    try:
        return a / b
    
    except ZeroDivisionError:
        return 'Cannot divide by zero'


safe_divide(10, 0)
```

<h3> 2.Simulate a Zomato-style rating system: ask the user for number of reviews and total stars, then calculate average rating. Use try-except to handle invalid (non-numeric) input and print an error message if input is not a number.<br><br><em><strong>Hint:</strong> Use input() and int() conversion inside a try block.</em>


```python
def calculate_zomato_rating():
    try:
        reviews = int(input("Enter Your value: "))
        stars = int(input("Enter Your value: "))
        
        average_rating = stars / reviews
        
        print(f"\n Zomato Average Rating: {average_rating} Stars")
        
    except ValueError:
        print("Error: Invalid input! ")

calculate_zomato_rating()
```

<h3>3.Create a function get_playlist_duration(songs) that takes a list of song durations (in seconds) and returns the total duration in minutes. Raise a custom exception InvalidDurationError if any duration in the list is negative.<br><br><em><strong>Hint:</strong> Define your own exception class by subclassing Exception.</em>

```python
class InvaliddurationError(Exception) :
    pass 

def get_playlist_duration(songs):
    total_seconds = 0
    for i in songs :
        if i < 0 :
            raise InvaliddurationError("Duration negative Value")
        total_seconds += i 

    total_minutes = total_seconds / 60
    return total_minutes


try:
    good_playlist = [180, -240, 200] 
    mins = get_playlist_duration(good_playlist)
    print("Playlist Total Duration:  ",mins)
except InvaliddurationError as e:
    print("Error caught: " , e)
```

<h3> 4.
Build a Flipkart-style order summary: ask the user for item price and quantity, then calculate and print total price. Use try-except-else-finally blocks to handle ValueError for invalid input, print the total if successful, and always print 'Thank you for shopping!' in the finally block.


```python
def flipkart_order_summary():   
    try:
        price = float(input("Enter item price (₹): "))
        quantity = int(input("Enter quantity: "))
        
    except ValueError:
        print("Error: Invalid input! Please enter valid numbers only.")
        
    else:
        total_price = price * quantity
        print(f"Order Successful! Your total amount is: ₹{total_price}")
        
    finally:
        print("Thank you for shopping with Flipkart!\n")

flipkart_order_summary()
```

```python
def smart_calculator():    
    try:
        num1 = int(input("Enter the first number: "))
        num2 = int(input("Enter the second number: "))
        result = num1 / num2
        print(f"Result: {num1} / {num2} = {result}")
        
    except ValueError:
        print("ValueError: Invalid input! Please enter only numbers.")
    except ZeroDivisionError:
        print("ZeroDivisionError: You cannot divide a number by zero.")

smart_calculator()
```

<h3>4.Given two sets: set1 contains the names of restaurants you have ordered from on Zomato, and set2 contains the names of restaurants you have ordered from on Swiggy, find and print the union and intersection of these sets.<br><br><em><strong>Hint:</strong> Use the union() and intersection() methods of Python sets.</em>

```python
Zomato = {"Dominos", "Burger King", "KFC", "Subway"}
Swigge = {"McDonalds", "KFC", "Pizza Hut", "Dominos"}

Union_restaurants = Zomato.union(Swigge)
Intersection = Zomato.intersection(Swigge)

print("Union")
print("-"*70)
print(Union_restaurants)
print("-"*70)
print("Swigge")
print(Intersection)
print("-"*70)
```

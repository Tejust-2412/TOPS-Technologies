```python
import pandas as pd
import numpy as np
data = pd.read_csv("sales_data_sample.csv")
```

<h4> 1. Identify the string methods and their execution order needed to remove extra
spaces and fix inconsistent casing in a product name.

To clean up messy text, we first use .strip() to remove leading and trailing white spaces, followed by .title() to ensure uniform capitalization. The execution order is crucial so that spaces are cleared before applying the text format

```python
data.columns = data.columns.str.strip().str.upper()
```

<h4> 2. Explain why manual CSV loading often treats numeric columns as strings and
identify the function used to convert them to floats.


CSV files store data as raw text, which often causes numeric columns containing special characters to be loaded as strings. We must clean these characters and use astype(float) or pd.to_numeric() to convert them into usable decimal numbers for calculations

<h4> 3. Determine the best data structure for mapping product names to total sales
and explain the drawback of using a list instead.


```python
for col in data.select_dtypes(include=['object']).columns:
        data[col] = data[col].str.strip()
```

```python
data['SALES'] = data['SALES'].replace({r'\$': '', ',': ''}, regex=True)
```

```python
data['SALES'] = pd.to_numeric(data['SALES'], errors='coerce').fillna(0.0)
```

<h4> 4. Compare using a manual for loop versus the sum() function, identifying a
scenario where the loop is the only viable option.


```python
summary = data.groupby('PRODUCTLINE')['SALES'].sum().reset_index()
```

<h4> 6. Explain the benefits of modularizing a data pipeline into separate functions
and the risks of using a single large code block in analytics.

```python
summary['CATEGORY'] = np.where(
        summary['SALES'] > 100000, 
        'High Revenue Product', 
        'Standard Revenue Product'
    )
```

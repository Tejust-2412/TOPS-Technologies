```python
def clean_brand_name(name):
    print(name.strip().replace("-"," "))

clean_brand_name(" oneplus-Nord ")
```

```python
string = "Apple iPhone 14 Pro Max"
string = string.split()
print("Brand Name :" ,string[0])
model = " ".join(string[1:])
print("Model Name :" ,model)
```

```python
def format_product_display(name, price):
    return f"{name} - {price}"

format_product_display("Boat earbuds" , 1299)
```

```python
product_names = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']
cleaned_names = []

for x in product_names :
    clean = x.strip().replace("-" , " ").title()
    cleaned_names.append(clean)


print(product_names)
print(cleaned_names)
```

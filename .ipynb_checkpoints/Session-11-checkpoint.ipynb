<h3>1.Write a recursive function in Python called reverse_string(s) that takes a string and returns it reversed (e.g., 'hello' becomes 'olleh').

```python
def reverse_string(s) :
    if len(s) == 0 :
        return s
    return reverse_string(s[1:]) + s[0]

s = "hello"
print(reverse_string(s))
```

<h3> 2.Build a recursive function sum_playlist_durations(durations) that takes a list of song durations (in seconds) and returns the total duration, similar to how Spotify totals a playlist.

```python
def sum_playlist_durations(durations):
    
    if len(durations) == 0:
        return 0
    
    return durations[0] + sum_playlist_durations(durations[1:])


playlist = [180, 240, 200, 150]
print(sum_playlist_durations(playlist))
```

<h3> 3.Given the following code, identify whether the variable 'count' is local or global in each function, and explain what will be printed when run:

```python
count = 10 
def update_count():
    count = 5
    print('Inside:', count)

update_count()
print('Outside:', count)
```

1.Inside: 5 is printed first. When you call update_count(), Python looks at the code inside the function. Inside the function, the value of count is set to 5.
<br>
2.Outside: 10 is printed next. After the function finishes its job, Python goes back to the normal code outside. Outside the function, the original count is still 10. The function did not change the outside number.

<h3> 4.Create a recursive function count_likes(posts) that takes a nested dictionary representing Instagram posts and their replies (each with a 'likes' key), and returns the total number of likes across all posts and replies.<br><br><em><strong>Hint:</strong> Each reply can itself have more replies, so use recursion to sum likes at all levels.</em>

```python
def count_likes(posts):
    total_likes = 0

    for post in posts:
        total_likes += post.get('likes', 0)
        
        if 'replies' in post:
            total_likes += count_likes(post['replies'])
            
    return total_likes

instagram_data = [
    {
        'text': 'Beautiful sunset!',
        'likes': 100,
        'replies': [
            {
                'text': 'Wow!', 
                'likes': 15, 
                'replies': []
            },
            {
                'text': 'Where is this?', 
                'likes': 5,
                'replies': [
                    {'text': 'Malibu', 'likes': 2, 'replies': []}
                ]
            }
        ]
    }
]

result = count_likes(instagram_data)
print(f"Total Likes Across Everything: {result}")
```

<h3> 5.Write a Python script that demonstrates the lifetime of a local variable inside a function versus a global variable by printing their values before, during, and after a function call. Use variable names similar to 'user_status' and 'app_status', inspired by WhatsApp online/offline status.

```python
app_status = "Connected to Server"
user_status = "Offline"
def open_whatsapp():
    user_status = "Online (Typing...)"
    chat_theme = "Dark Mode"
    
    print("--- During the funtion call ---")
    print("App Status (Global) : ", app_status)
    print("User Status (Local) : ",user_status)
    print("Chat Theme (Local)  : " ,chat_theme)

print("--- Before the function call ---")
print("App Status : ",app_status)
print("User Status: ",user_status)
open_whatsapp()
print("--- After the function call ---")
print("App Status : ",app_status)
print("User Status: ",user_status)
```

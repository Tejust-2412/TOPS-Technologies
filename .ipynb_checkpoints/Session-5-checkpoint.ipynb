<h3> 2.Simulate a Flipkart shopping cart: start with a list cart_items containing 't-shirt', 'shoes'. Use extend() to add ['jeans', 'cap'] to the cart, then print the final list of items.

```python
cart_items = ['T-shirt' , 'Shoes']
cart_items.extend(['Jeans' , 'Cap'])
print("Final List Item : ",cart_items)
```

<h3> 3.Write a function remove_last_item(order_list) that pops the last item from a Zomato order list and returns the removed item. Test it with a sample order_list.

```python
def remove_last_item(order_list) :
    return order_list.pop()

Zomato_order_list = ["Cake" , "Burger" , "Pizza" , "Maggie"]
remove_last_item(Zomato_order_list)
```

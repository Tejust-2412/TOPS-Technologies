<h2> 2) Write a Python program that takes a user's input for the price of a Zomato order as a string, converts it to a float using type casting, adds 18% GST, and prints the final bill amount.

```python
Price = str(input("Enter Your Price : "))
Price = float(Price)

GST_amount = Price*0.18
Final_amount= Price + GST_amount
print(Final_amount)
```

<h2> 3) Given a list of strings representing product prices from Flipkart, like ['199.99', '299.50', '150'], convert all to floats and calculate the total cart value.

```python
Prices = ['199.99', '299.50', '150']
Converted_Price = []
cart_values = 0
for i in Prices :
    Float_value = float(i)
    Converted_Price.append(Float_value)
    cart_values += Float_value

print(Converted_Price)
print(sum(Converted_Price))
```

<h2> 4)Build a function is_discount_applicable(order_amount) that takes a float and returns True if the amount is greater than 500, otherwise False. Print the result for order amounts 450 and 750.

```python
def is_discount_applicable(order_amount) :
    float(order_amount)
    if order_amount >= 500 :
        return True
    else :
        return False


is_discount_applicable(450.50)
is_discount_applicable(700)
```

<h2> 5.You received a dataset of ratings as strings from Spotify: ['4.5', '3.0', '5', '4.2']. Use type casting to convert these to floats, then find and print the highest rating.<br><br><em><strong>Hint:</strong> Use the float() function inside a loop or list comprehension.</em>

```python
Spotify = ['4.5' , '3.0' , '5' , '4.2']
floar_number = [float(i) for i in Spotify]

print(max(floar_number))
```

<h3>1.Define a function called calculate_final_price(price, discount_rate) that returns the final price after applying the discount. Test it with price 1200 and discount_rate 0.15.


```python
def calculate_final_price (price, discount_rate) :
    discount = price*discount_rate
    total = price - discount
    return total 


calculate_final_price(1200 , 0.15)
```

<h3> 2.Create a function get_delivery_charge(amount, city='Ahmedabad') that returns a delivery charge: Rs. 30 for Ahmedabad, Rs. 50 for other cities. Call it with and without the city argument to see both results.


```python
def get_delivery_charge(amount, city='Ahmedabad'):
    return (f"Rs.{amount} for {city}" )

get_delivery_charge(30)
```

```python
get_delivery_charge(50, "surat")
```

<h3> 3.Build a function called format_coupon_message(username, discount=10) that returns a string like 'Hi Rahul, you get 10% off!' If no discount is given, use 10% by default. Test it for two users: one with a custom discount, one with the default.

```python
def format_coupon_message(username, discount=10) :
    return (f"Hi {username}, you get {discount}% off! ")

format_coupon_message("Tejas")
```

```python
format_coupon_message("Nikhil" , 5)
```

<h3> 4.Given a function apply_discount(price, rate=0.10), update it so that if the rate is not passed, it uses 0.10 by default. Then, call it with only the price argument and print the result.<br><br><em><strong>Hint:</strong> Use default arguments in your function definition.</em>

```python
def apply_discount(price, rate=0.10):
    discount = price*rate
    total = price - discount
    return total 

apply_discount(1400)
```

<h3>5.Write a function called calculate_cashback(amount, cashback_rate=0.05) that returns the cashback amount. Then, use it to calculate cashback for a Zomato order of Rs. 500 with the default rate, and for a Flipkart order of Rs. 2000 with a 7% cashback.

```python
def calculate_cashback(amount, cashback_rate=0.05) :
    discount = amount * cashback_rate
    return discount

print("Zomato order : " ,calculate_cashback(500))
```

```python
print("Flipkart order : " ,calculate_cashback(2000 , 0.07))
```

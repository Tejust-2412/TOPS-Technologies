<h3> 3.Given two lists — one of food items and one of prices — use zip() to print each food item with its price like a Zomato menu (e.g., 'Pizza - ₹250').

```python
food_items = ["Margherita Pizza", "Aloo Tikki Burger", "White Sauce Pasta", "Peri Peri Fries", "Chocolate Brownie"]
prices = [250, 99, 199, 120, 150]
for food, price in zip(food_items, prices):
    print(f"{food} - ₹{price}")
```

<h3> 4.Write a generator function called insta_posts_generator(posts) that takes a list of Instagram post captions and yields one caption at a time. Use next() to get and print the next post caption each time until all captions are printed.<br><br><em><strong>Hint:</strong> Use the yield keyword inside your function and handle StopIteration when all posts are done.</em>

```python
def insta_posts_generator(posts):
    for post in posts:
               yield post


captions_list = [
    "Enjoying the weekend vibes! ", 
    "Beautiful sunset at the beach ", 
    "Workout completed! ",
    "New setup alert!"
]
feed = insta_posts_generator(captions_list)

while True:
    try:
        current_post = next(feed)
        print(f"{current_post}")
        
    except StopIteration:
        print("You're all caught up! No more posts to show.")
        break
```

<h3> 5.Build a generator function called cashback_generator(transactions) that takes a list of Paytm transaction amounts and yields 5% cashback for each transaction. Print out the cashback values for all transactions.

```python
def cashback_generator(transactions):
    for amount in transactions:
        cashback = amount * 0.05
        yield cashback


paytm_transactions = [200, 1500, 50, 400, 1000]

cashback_feed = cashback_generator(paytm_transactions)

for amount, cashback_earned in zip(paytm_transactions, cashback_feed):
    print(f"Transaction: ₹{amount:4} | Cashback Earned: ₹{cashback_earned}")
```

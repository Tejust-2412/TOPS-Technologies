```python
price = [100,250,500]
a = list(map(lambda i : i+i*0.18  , price))
print(a)
```

```python
favorite_songs = ['KeSariya', 'BelieVer', 'Shape-OF-You', 'Blinding LIghts', 'Excuses']

li = list(map(lambda favorite_songs: favorite_songs.strip().title().replace("-" , " "), favorite_songs))
```

<h3> 4.Given a list of order amounts from a Zomato cart [120, 340, 560, 80], use reduce() from functools to calculate the total bill amount.

```python
from functools import reduce 
cart = [120, 340, 560, 80]
a =  reduce(lambda a,b : a+b , cart)
a
```

<h3> 5.Use ChatGPT or Copilot to generate a Python code snippet that uses map(), filter(), and reduce() together to process a list of numbers: first double each number, then filter to keep only numbers greater than 100, and finally sum the result. Paste and test the generated code with the list [40, 60, 80, 120].

```python
from functools import reduce
numbers = [40, 60, 80, 120]
doubled_numbers = list(map(lambda x: x * 2, numbers))
filtered_numbers = list(filter(lambda x: x > 100, doubled_numbers))
final_result = reduce(lambda x, y: x + y, filtered_numbers)

print("Original Numbers:        ", numbers)
print("After map (Doubled):     ", doubled_numbers)
print("After filter (> 100):    ", filtered_numbers)
print("After reduce (Final Sum):", final_result)
```

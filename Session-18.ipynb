<h3> 1.Create a Python class called Product with a private attribute _price. Initialize _price in the constructor and write a method to display its value.


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price 
        
    def display_price(self):
        print(f"The price of '{self.name}' is ₹{self._price}")


my_product = Product("Wireless Headphones", 1999)
my_product.display_price()
```

<h3> 2. Add getter and setter methods for the _price attribute in your Product class to safely access and update the price. Make sure the setter prevents setting a negative price.<br><br><em><strong>Hint:</strong> Raise a ValueError if the new price is less than zero.</em>


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = 0 
        self.set_price(price) 
        
    def get_price(self):
        return self._price
        
    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Error: Price cannot be negative!")
        else:
            self._price = new_price
            print(f"Price updated successfully to ₹{self._price}")


my_product = Product("Wireless Headphones", 1999)
print("Updating price to ₹1500...")
my_product.set_price(1500)

try:
    my_product.set_price(-500)
except ValueError as e:
    print(e) 
```

<h3> 3.Build a class called Playlist that has a private attribute _songs (a list of song names). Write methods to add a song, remove a song, and get the current list of songs using proper encapsulation.


```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self._songs = [] 
        
    def add_song(self, song_name):
        self._songs.append(song_name)
        print(f"Added to'{self.name}': {song_name}")
        
    def remove_song(self, song_name):
        if song_name in self._songs:
            self._songs.remove(song_name)
            print(f"Removed from '{self.name}': {song_name}")
        else:
            print(f"Error: '{song_name}' is not in the playlist!")
            
    def get_songs(self):
        return self._songs.copy()


my_playlist = Playlist("Workout Mix")

my_playlist.add_song("Still D.R.E")
my_playlist.add_song("Roddy Ricch")
my_playlist.add_song("Believer")

print("Current Playlist:", my_playlist.get_songs())

my_playlist.remove_song("Lose Yourself")
print("Updated Playlist:", my_playlist.get_songs())

my_playlist.remove_song("Shape of You")
```

<h3> 4.Create an abstract class PaymentMethod with an abstract method pay(amount). Then, create two subclasses: UPI and CreditCard, each implementing the pay method with a print statement showing how the payment would be processed.<br><br><em><strong>Hint:</strong> Use the abc module for abstraction.</em>


```python
from abc import ABC, abstractmethod
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass 

class UPI(PaymentMethod):
    def __init__(self, upi_id):
        self.upi_id = upi_id
    
    def pay(self, amount):
        print(f"Processing ₹{amount} securely via UPI: {self.upi_id}")

class CreditCard(PaymentMethod):
    def __init__(self, card_number):
        self.card_number = card_number
        
    def pay(self, amount):
        last_4_digits = self.card_number[-4:]
        print(f"Processing ₹{amount} via Credit Card ending in {last_4_digits}")

my_gpay = UPI("user@okhdfcbank")
my_visa = CreditCard("4111222233334444")

my_gpay.pay(1500)
my_visa.pay(4500)
```
